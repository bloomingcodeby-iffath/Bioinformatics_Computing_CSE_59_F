##### 13.Touple


```python
color=("red","green","blue","yellow","black")
print(color)
#1st color
print(color[0])
#last color
print(color[len(color)-1]),
#count red
print(color.count("red"))
#index of blue
print(color.index("blue"))
#1st 3 colors
print(color[0:3])
#last 3
print(color[len(color)-3:len(color)])
#length
print(len(color))
#reverse
reverse=color[::-1]
print(reverse)
```

##### Recursion

```python
#sum of n numbers
def sum(n):
    if(n==0):
        return 0
    return n + sum(n-1)
sum(5)
```

```python
# power
def power(b,n):
    if (n==0):
        return 1
    return b * power(b,n-1)
power(3,3) 
```
