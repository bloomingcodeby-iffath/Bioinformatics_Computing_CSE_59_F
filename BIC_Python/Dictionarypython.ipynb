##### 1. Student-এর তথ্য নিয়ে Dictionary তৈরি করো

```python
name=input("Name:")
id=input("id:")
info={
    "Name":name,
    "ID":id
}
info.update({
    "Dept": "CSE",
    "Batch": 59
})
for key,value in info.items():
    print(key, ":",value)
```

#### 1. String-এর প্রতিটি character কতবার আছে

```python
text="Iffath Sanjida"
count={}
for ch in text:
    if(ch!=" "):
        if(ch in count):
            count[ch]+=1
        else:
            count[ch]=1
count
```

#### 2. String-এর প্রতিটি word কতবার আছে

```python
string="apple banana apple mango banana apple"
s2=string.split()
dict={}
for s in s2:
    if(s in dict):
        dict[s]+=1
    else:
        dict[s]=1
dict
```

##### 3. String-এ প্রতিটি vowel কতবার আছে

```python
def function(msg):
    vowels="aeiou"
    countVowel={}
    for ch in msg:
     if ch in vowels:
        if ch in countVowel:
            countVowel[ch]+=1
        else :
            countVowel[ch]=1

    for value in countVowel.items():
        print(value)
msg="aaeioeeuuaaiieuoaie"
function(msg)
```
