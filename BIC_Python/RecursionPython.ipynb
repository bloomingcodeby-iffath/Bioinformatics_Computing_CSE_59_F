### 1. 1 থেকে N পর্যন্ত print

```python
def number(n):
    if(n<=0):
        return 0
    if(n%2==0):
        return n + number+(n-1)
    else:
        return number(n-1)
number(5)
```

### 2. N থেকে 1 পর্যন্ত print

```python
def numberPrint(n):
    if(n<0):
        return 0
    print(n)
    return numberPrint(n-1)

numberPrint(5)
```

### 3. 1 থেকে N পর্যন্ত যোগফল

```python
def numberSum(n):
    if(n==0):
        return 0
    return n + numberSum(n-1)
numberSum(5)
```

### 4. N-এর factorial বের করো

```python
def factorial(n):
    if(n==1) or (n==0):
        return 1
    return n*factorial(n-1)
factorial(4)
```

#### 5. দুইটি সংখ্যার power বের করো

```python
def power(base,p):
    if(p==0):
        return 1
    return base * power(base,p-1)
power(2,2)
```

#### 6. একটি সংখ্যার digit-এর sum বের করো

```python
def digitSum(n):
    if(n==0):
        return 0
    return (n%10)+digitSum(n//10)
digitSum(12345)
```

#### 7. Digit কতটি?

```python
def countDigit(n):
    if(n==0):
        return 0
    return 1+countDigit(n//10)
countDigit(100)
```

#### 8. Number Reverse

```python
def reverse(s):
    if(len(s)==0):
        return ""
    return reverse(s[1:])+s[0]
reverse("iffat")
```

#### 10. Fibonacci N-th Term

```python
def Fibo(n):
    if(n==0):
        return 0
    if(n==1):
        return 1
    return Fibo(n-1)+Fibo(n-2)
Fibo(7)
```

#### 12. Odd digit count

```python
def countOdd(n):
    if(n==0):
        return 0

    if (n % 10) % 2 != 0:
        return 1 + countOdd(n // 10)
    else:
        return countOdd(n // 10)

countOdd(103)
```

```python
def evenSum(n):
    if(n==0):
        return 0
    if(n%10) % 2 ==0:
        return (n%10)+evenSum(n//10)
    else:
        return evenSum(n//10)

evenSum(444)    
```
