```python
#Palindrome
def Palindrome(s1):
    if(s1==s1[::-1]):
        print("its a palindrome!")
    else:
        print("not a palindrome")
Palindrome("AAbAAs")
```

```python
#Anagram: 
def anagram(str1,str2):
    if(sorted(str1)==sorted(str2)):
        print("Anagram")
    else:
        print("Not an anagram")
anagram("heart","earth")
```

```python
#uppercase lowercase count
def uplow(s):
    lower=0
    upper=0
    for ch in s:
        if(ch.islower()):
            lower+=1
        if(ch.isupper()):
            upper+=1
    return upper,lower
uplow("Hello World")
```
