```python
num=[-1,2,-9,4,-3,5,4,3]
pos=[]
neg=[]
for n in num:
    if(n>0):
        pos.append(n)
    else:
        neg.append(n)
print(pos)
print(neg)
```

```python
numbers=[1,2,3,4,5]
even=[]
odd=[]
evenSum=0
for n in numbers:
    if n%2==0:
        evenSum+=n
        even.append(n)
    else:
        odd.append(n)

print(even)
print(odd)
print("even sum",evenSum)
print("Maximum: ",max(numbers))
print("Minimum: ",min(numbers))
```
