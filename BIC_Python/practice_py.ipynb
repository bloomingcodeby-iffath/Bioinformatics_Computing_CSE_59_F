```python
text = "hello world"

count = {}

for ch in text:
    if ch != " ":
        if ch in count:
            count[ch] += 1
        else:
            count[ch] = 1

print(count)
```

```python
List=["apple","apple","lichi","banana","banana","banana"]
countFruits={}

for i in range(len(List)):
    if List[i] in countFruits:
        countFruits[List[i]]+=1
    else:
        countFruits[List[i]]=1

countFruits
```

```python
def Vowel(str):
    count=0
    for ch in str:
        if ch.lower() in "aeiou":
            count+=1
    return count
print(Vowel("Hello World"))
```

```python
num=[1,2,3,4,5]
num2=[]
num3=[]
print("Even Number")
for n in num:
    if n%2==0:
        num2.append(n)
print(num2)
print("Odd Number")
for n in num:
    if n%2!=0:
        num3.append(n)
print(num3)
```

```python
def isAnagram(s1,s2):
    if(sorted(s1)==sorted(s2)):
        return 1
    else:
        return 0
print(isAnagram("earth","heart"))
```
