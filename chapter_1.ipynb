#### *(1A) Compute the Number of Times a Pattern Appears in a Text*
##### Input:
      Text → a string of nucleotides (A, T, G, C)
      Pattern → a shorter string to search for inside Text
##### Output:
A single integer — the number of times Pattern appears in Text

#### Solution:

Use a sliding window of size len(Pattern) across Text.  
At every position i, extract a substring of the same length as Pattern and compare. If they match, increment the counter.

Text    =  G C G C G  
Position:  0 1 2 3 4

i=0 → "GCG" == "GCG" ✅  count = 1  
i=1 → "CGC" != "GCG" ❌  
i=2 → "GCG" == "GCG" ✅  count = 2 

```python
def pattern_count(text, pattern):
    count = 0               
    k = len(pattern)    
    for i in range(len(text) - k + 1):
        if text[i:i+k] == pattern:  
            count += 1
    return count

# input:
text    = input("Enter Text: ").strip()
pattern = input("Enter Pattern: ").strip()

# output
print(pattern_count(text, pattern))
```

#### *1B — Find the Most Frequent Words in a String*
##### 📥 Input  
Text: ACGTTGCATGTCGCATGATGCAT  
k:    4  
##### 📤 Output  
GCAT CATG  
All k-mers that appear the maximum number of times in Text.  

💡 Approach  

Use a frequency dictionary (hash map):  

- Slide a window of size k across Text  
- Count how many times each k-mer appears  
- Find the maximum count  
- Return all k-mers that have that maximum count  

Text = ACGTTGCATGTCGCATGATGCAT
k    = 4

k-mers and their counts:  
ACGT → 1  
CGTT → 1  
GTTG → 1  
TTGC → 1  
TGCA → 2  ✅  
GCAT → 3  ✅ (max)  
CATG → 3  ✅ (max)  
ATGT → 1  
...  
  
Most frequent → GCAT, CATG (both appear 3 times)  

```python
def frequent_words(text, k):
    freq = {}  # dictionary to store k-mer frequencies

    # count every k-mer in the text
    for i in range(len(text) - k + 1):
        kmer = text[i:i+k]           # extract k-mer at position i
        freq[kmer] = freq.get(kmer, 0) + 1  # increment its count

    max_count = max(freq.values())   # find the highest frequency

    # collect all k-mers that have the maximum frequency
    result = [kmer for kmer, count in freq.items() if count == max_count]

    return result

# ---------- Input ----------
text = input("Enter Text: ").strip()
k    = int(input("Enter k: ").strip())

# ---------- Output ----------
print(*frequent_words(text, k))
```
